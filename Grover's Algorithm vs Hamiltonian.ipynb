```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

# Define the equation we want to solve for A
def equation_for_A(A, theta):
    B = np.sqrt(1 - A**2)  # Since |A|^2 + |B|^2 = 1
    return A * B * np.sin(2 * theta) + A**2 * np.cos(2 * theta) - np.cos(theta)**2

# Set up theta values
theta_vals = np.linspace(0, 2 * np.pi, 1000)
A_vals = []

# Solve for A for each theta value
for theta in theta_vals:
    # Use fsolve to find the root of the equation
    A_initial_guess = 0.5  # Initial guess for |A|
    A_solution = fsolve(equation_for_A, A_initial_guess, args=(theta))[0]
    A_vals.append(A_solution)

# Convert to numpy array for easier manipulation
A_vals = np.array(A_vals)

# Find the max and min values of A and their corresponding theta values
A_max = np.max(A_vals)
A_min = np.min(A_vals)

theta_max = theta_vals[np.argmax(A_vals)]
theta_min = theta_vals[np.argmin(A_vals)]

# Print max and min values and their corresponding theta
print(f"Max value of |A|: {A_max} at θ = {theta_max} radians")
print(f"Min value of |A|: {A_min} at θ = {theta_min} radians")

# Plot the results
plt.plot(theta_vals, A_vals, label="|A| vs θ")
plt.xlabel('θ (radians)')
plt.ylabel('|A|')
plt.title('|A| vs θ')
plt.scatter([theta_max, theta_min], [A_max, A_min], color='red', label='Min/Max Points')
plt.legend()
plt.grid(True)
plt.show()
```

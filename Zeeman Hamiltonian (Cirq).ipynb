```python
import cirq
```

```python
# Initilizing the qubits
ancil = cirq.NamedQubit("ancil")
ancil_sub = cirq.NamedQubit("ancil_sub")
ancil_reset = cirq.NamedQubit("ancil_reset")
q1 = cirq.NamedQubit("q1")
q2 = cirq.NamedQubit("q2")
q3 = cirq.NamedQubit("q3")
q4 = cirq.NamedQubit("q4")
q5 = cirq.NamedQubit("q5")
```

```python
circuit = cirq.Circuit(
    # Applying Hadamard Gate to all qubits
    cirq.H(ancil),
    cirq.H(ancil_sub),
    cirq.X(ancil_reset),
    cirq.H(q1),
    cirq.H(q2),
    cirq.H(q3),
    cirq.H(q4),
    cirq.H(q5),
    
    # Applying 4-Qubit Toffoli Gate
    
    ### On q1,q2,q3
    cirq.TOFFOLI(q1, q2, ancil_sub),
    cirq.TOFFOLI(ancil_sub, q3, ancil),
    
    ### On q1,q2,q4
#     cirq.TOFFOLI(q1, q3, ancil_sub),
#     cirq.TOFFOLI(ancil_sub, q4, ancil),
#     cirq.X(ancil_sub),
    
#     ### On q1,q2,q5
#     cirq.TOFFOLI(q1, q3, ancil_sub),
#     cirq.TOFFOLI(ancil_sub, q5, ancil),
#     cirq.X(ancil_sub),
    
    # Measurement
#     cirq.measure(ancil),
#     cirq.measure(q1),
#     cirq.measure(q2),
#     cirq.measure(q3),
#     cirq.measure(q4),
#     cirq.measure(q5)
)
```

```python
print(circuit)
```

```python
print("\nMoments in the circuit:\n")
for i, moment in enumerate(circuit):
    print('Moment {}: {}'.format(i,moment))
```

```python
simulator = cirq.Simulator()
```

```python
result = simulator.run(circuit, repetitions=5)
```
